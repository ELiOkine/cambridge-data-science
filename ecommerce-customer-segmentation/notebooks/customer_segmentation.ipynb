```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE
from scipy.cluster.hierarchy import linkage, dendrogram

# Just setting a calm default style
sns.set(style="whitegrid")
```

```python
# Pull the data from the course link
sales_raw_frame = pd.read_csv(
    "https://github.com/FourthRevGitHub/CAM_DS_Public-Data-Sets/raw/main/Course%201/Week%206/CUSTOMERS_CLEAN.zip"
)
```

# First few rows to see the layout
sales_raw_frame.head()

### Part 2: Column clean-up and core fields

I want the column names consistent before doing any feature work.

```python

cleaned_sales_frame = sales_raw_frame.copy()

cleaned_sales_frame.columns = (
    cleaned_sales_frame.columns.str.strip()
    .str.lower()
    .str.replace(" ", "_", regex=False)
    .str.replace("/", "_", regex=False)
    .str.replace("-", "_", regex=False)
)
```

```python
# Column list after cleanup
cleaned_sales_frame.columns
```

```python


def pick_column_name(candidates, available_columns):
    for column_name in candidates:
        if column_name in available_columns:
            return column_name
    raise KeyError(f"None of the columns found: {candidates}")

customer_id_column = pick_column_name(["customer_id", "customer_id_"], cleaned_sales_frame.columns)
order_id_column = pick_column_name(["order_id"], cleaned_sales_frame.columns)
order_date_column = pick_column_name(["order_date"], cleaned_sales_frame.columns)
delivery_date_column = pick_column_name(["delivery_date"], cleaned_sales_frame.columns)
revenue_column = pick_column_name(["total_revenue", "total_revenue_usd"], cleaned_sales_frame.columns)
unit_cost_column = pick_column_name(["unit_cost"], cleaned_sales_frame.columns)
profit_column = pick_column_name(["profit", "total_profit"], cleaned_sales_frame.columns)
quantity_column = pick_column_name(["quantity"], cleaned_sales_frame.columns)
birthdate_column = pick_column_name(["customer_birthdate", "customer_birth_date", "birthdate"], cleaned_sales_frame.columns)
```

```python
# Some money fields are strings, so I’m cleaning them up here
numeric_columns_to_clean = [revenue_column, unit_cost_column, profit_column, quantity_column]

for column_name in numeric_columns_to_clean:
    cleaned_sales_frame[column_name] = (
        cleaned_sales_frame[column_name]
        .astype(str)
        .str.replace(r"[^0-9.\-]", "", regex=True)
        .replace("", pd.NA)
        .astype(float)
    )
```

```python
# Date parsing so later calculations behave
for date_column in [order_date_column, delivery_date_column, birthdate_column]:
    cleaned_sales_frame[date_column] = pd.to_datetime(
        cleaned_sales_frame[date_column], errors="coerce", format="mixed"
    )
```

#### Analysis

Column names are now uniform, which should prevent small mistakes later. The core fields are identified up front, so the rest of the notebook can stay readable.

### Part 3: Data quality checks

A few focused checks before building customer-level features.

```python
# Quick check to see how bad the missing data is
missing_value_counts = cleaned_sales_frame.isna().sum()
missing_value_counts[missing_value_counts > 0].sort_values(ascending=False).head(10)
```

```python
# Nothing unusual here, but good to confirm
duplicate_row_count = cleaned_sales_frame.duplicated().sum()
duplicate_row_count
```

```python
# This helps spot any obvious outliers early
numeric_columns_for_outliers = [quantity_column, revenue_column, unit_cost_column, profit_column]

outlier_counts = {}
for column_name in numeric_columns_for_outliers:
    value_series = pd.to_numeric(
        cleaned_sales_frame[column_name], errors="coerce"
    ).dropna()
    lower_quartile = value_series.quantile(0.25)
    upper_quartile = value_series.quantile(0.75)
    interquartile_range = upper_quartile - lower_quartile
    lower_bound = lower_quartile - 1.5 * interquartile_range
    upper_bound = upper_quartile + 1.5 * interquartile_range
    outlier_counts[column_name] = int(
        ((value_series < lower_bound) | (value_series > upper_bound)).sum()
    )

pd.Series(outlier_counts).sort_values(ascending=False)
```

#### What I See

Missing values show up in a few places, which will matter when we build customer features. The outlier counts give a rough sense of heavy tails, but I’ll handle it with scaling rather than manual trimming.

### Part 4: Build customer-level features

Now I’ll collapse the orders into one row per customer.

```python
# Reference date for recency and age calculations
reference_date = cleaned_sales_frame[order_date_column].max() + pd.Timedelta(days=1)
```

```python
# One row per customer
customer_feature_table = (
    cleaned_sales_frame.groupby(customer_id_column)
    .agg(
        frequency=(order_id_column, "nunique"),
        last_order_date=(order_date_column, "max"),
        clv=(revenue_column, "sum"),
        avg_unit_cost=(unit_cost_column, "mean"),
        customer_birthdate=(birthdate_column, "first"),
    )
    .reset_index()
)
```

```python
# Recency and age features
customer_feature_table["recency"] = (
    reference_date - customer_feature_table["last_order_date"]
).dt.days
customer_feature_table["customer_age"] = (
    (reference_date - customer_feature_table["customer_birthdate"]).dt.days / 365.25
)
```

```python
# Quick guardrail for odd ages
customer_feature_table.loc[
    (customer_feature_table["customer_age"] < 0)
    | (customer_feature_table["customer_age"] > 100),
    "customer_age",
] = pd.NA
```

```python
# Keep only the engineered features plus ID
customer_feature_table = customer_feature_table.drop(
    columns=["last_order_date", "customer_birthdate"]
)
```

```python
# Quick peek at the new table
customer_feature_table.head()
```

#### Analysis

The customer table is compact and focused on the five features I need. Age has a few odd values, which is expected in real data, so I’ll rely on imputation later.

### Part 5: Descriptive stats and quick EDA

This is a light scan to see the overall shape before clustering.

```python
# Baseline stats for the five features
feature_cols = ["frequency", "recency", "clv", "avg_unit_cost", "customer_age"]
customer_feature_table[feature_cols].describe().T
```

Mean and median are both listed in the table. The gap between them hints at skew, especially for CLV and frequency.

```python
# Feature distributions

plot_figure, plot_axes = plt.subplots(2, 3, figsize=(15, 8))
plot_axes = plot_axes.flatten()

feature_labels = {
    "frequency": ("Purchase frequency per customer", "Number of orders"),
    "recency": ("Days since last order", "Days"),
    "clv": ("Customer lifetime value", "Total revenue (USD)"),
    "avg_unit_cost": ("Average unit cost", "Average unit cost (USD)"),
    "customer_age": ("Estimated customer age", "Years"),
}

for plot_index, feature_name in enumerate(feature_cols):
    plot_title, x_label = feature_labels[feature_name]
    sns.histplot(
        customer_feature_table[feature_name],
        kde=True,
        ax=plot_axes[plot_index],
        color="slategray",
    )
    plot_axes[plot_index].set_title(plot_title)
    plot_axes[plot_index].set_xlabel(x_label)
    plot_axes[plot_index].set_ylabel("Customer count")

plot_axes[-1].axis("off")
plt.tight_layout()
```

```python
# Correlation heatmap

plt.figure(figsize=(8, 6))
correlation_view = customer_feature_table[feature_cols].corr()
sns.heatmap(correlation_view, annot=True, cmap="viridis", fmt=".2f")
plt.title("How the features move together")
plt.xlabel("Feature group")
plt.ylabel("Feature group")
plt.show()
```

#### Observation

### Part 6: Scale and impute for clustering

I need a clean numeric matrix before any clustering step.

```python
# Scale and impute the feature table
numeric_transformer = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler()),
    ]
)

feature_preprocessor = ColumnTransformer(
    transformers=[("numeric", numeric_transformer, feature_cols)],
    remainder="drop",
)

scaled_feature_matrix = feature_preprocessor.fit_transform(customer_feature_table)
scaled_feature_matrix.shape
```

#### Analysis

The scaled matrix is ready and avoids issues from different units. This should make the clustering less sensitive to raw scale.

### Part 7: Pick the number of clusters

I’m using elbow and silhouette together so the choice isn’t just visual.

```python
# Compute elbow and silhouette values
inertia_scores = []
silhouette_scores = []
cluster_candidates = range(2, 11)

for cluster_count in cluster_candidates:
    kmeans_model = KMeans(n_clusters=cluster_count, random_state=42, n_init="auto")
    segment_labels = kmeans_model.fit_predict(scaled_feature_matrix)
    inertia_scores.append(kmeans_model.inertia_)
    silhouette_scores.append(silhouette_score(scaled_feature_matrix, segment_labels))
```

```python
# Visual check for the elbow and silhouette trends
plot_figure, plot_axes = plt.subplots(1, 2, figsize=(12, 4))

plot_axes[0].plot(cluster_candidates, inertia_scores, marker="o", color="teal")
plot_axes[0].set_title("Elbow curve for cluster count")
plot_axes[0].set_xlabel("Number of clusters")
plot_axes[0].set_ylabel("Within-cluster sum of squares")

plot_axes[1].plot(cluster_candidates, silhouette_scores, marker="o", color="#6A4C93")
plot_axes[1].set_title("Silhouette trend across k")
plot_axes[1].set_xlabel("Number of clusters")
plot_axes[1].set_ylabel("Silhouette score")

plt.tight_layout()
```

```python
# A rough elbow guess using the change in slope
second_differences = np.diff(inertia_scores, 2)
elbow_cluster_count = list(cluster_candidates)[
    int(np.argmax(np.abs(second_differences)) + 2)
]

# Silhouette peak
best_cluster_count = list(cluster_candidates)[int(np.argmax(silhouette_scores))]

elbow_cluster_count, best_cluster_count
```

#### Analysis

The elbow estimate and the silhouette peak are in the same ballpark, which is a good sign. I’ll use the silhouette pick as the final k unless the dendrogram really disagrees.

### Part 8: Hierarchical clustering check

This is a rough cross-check before finalising k-means.

```python
# A small sample keeps the dendrogram readable
sample_size = min(5000, scaled_feature_matrix.shape[0])
np.random.seed(42)
sample_index = np.random.choice(
    scaled_feature_matrix.shape[0], size=sample_size, replace=False
)
scaled_sample = scaled_feature_matrix[sample_index]

linked_matrix = linkage(scaled_sample, method="ward")

plt.figure(figsize=(12, 4))
dendrogram(
    linked_matrix,
    truncate_mode="lastp",
    p=30,
    link_color_func=lambda _: "slategray",
)
plt.title("Dendrogram check on a sample")
plt.xlabel("Cluster size (sample)")
plt.ylabel("Merge distance")
plt.show()
```

```python


dendrogram_cluster_count = best_cluster_count
dendrogram_cluster_count
```

#### Observation

The dendrogram spacing supports a small number of clusters rather than something huge. It lines up with the silhouette suggestion, which is reassuring.

### Part 9: K-means segmentation

Now I’ll fit k-means with the chosen cluster count.

```python
# Fit the k-means model
segment_count = best_cluster_count
kmeans_model = KMeans(n_clusters=segment_count, random_state=42, n_init="auto")
segment_labels = kmeans_model.fit_predict(scaled_feature_matrix)
```

```python
# Add segments back to the customer table
customer_feature_table["segment"] = segment_labels
```

```python
# Map of customer IDs to segments
customer_segment_table = customer_feature_table[[customer_id_column, "segment"]]
customer_segment_table
```

#### Analysis

The segments are now attached to each customer. From here I can profile the groups and see what actually separates them.

### Part 11: 2D views with PCA and t-SNE

These plots help check if segments separate in a compact view.

```python
# t-SNE (sample for efficiency)

tsne_sample_size = min(5000, scaled_feature_matrix.shape[0])
np.random.seed(42)
tsne_sample_index = np.random.choice(
    scaled_feature_matrix.shape[0], size=tsne_sample_size, replace=False
)

scaled_tsne_sample = scaled_feature_matrix[tsne_sample_index]
segment_sample_labels = segment_labels[tsne_sample_index]

tsne_model = TSNE(n_components=2, random_state=42, init="pca", learning_rate="auto")
tsne_coordinates = tsne_model.fit_transform(scaled_tsne_sample)

tsne_view_frame = pd.DataFrame(tsne_coordinates, columns=["TSNE1", "TSNE2"])
tsne_view_frame["segment"] = segment_sample_labels

plt.figure(figsize=(8, 6))
sns.scatterplot(
    data=tsne_view_frame,
    x="TSNE1",
    y="TSNE2",
    hue="segment",
    palette="viridis",
    s=20,
    alpha=0.6,
)
plt.title("t-SNE view of customer segments (sample)")
plt.xlabel("t-SNE axis 1")
plt.ylabel("t-SNE axis 2")
plt.legend(title="Segment", bbox_to_anchor=(1.02, 1), loc="upper left")
plt.show()
```
